```python
from tkinter import *
from tkinter import ttk

coeff_UE1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
coeff_UE2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
coeff_UE3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

def calcul_moyenne(notes, coeff):
    somme_notes = 0
    somme_coeff = 0
    for i in range(len(notes)):
        somme_notes += notes[i] * coeff[i]
        somme_coeff += coeff[i]
    return round(somme_notes / somme_coeff, 2)

def decision(moyennes):
    nb_valides = 0
    for m in moyennes:
        if m >= 10:
            nb_valides += 1
    if nb_valides == 3:
        return "Année validée, passage en BUT 2"
    elif nb_valides == 2 and min(moyennes) >= 8:
        return "Année non validée, passage en BUT 2 autorisé (UE à valider en BUT 2)"
    elif nb_valides == 2:
        return "Année non validée, passage non autorisé (redoublement possible)"
    else:
        return "Année non validée, passage en BUT 2 exclu"

def cal():
    try:
        notes = [float(saisie.get()) for saisie in saisies]
    except ValueError:
        resultat.config(text="Erreur : saisis un nombre dans chaque case")
        return
    moyennes = [calcul_moyenne(notes, coeff_UE1),
                calcul_moyenne(notes, coeff_UE2),
                calcul_moyenne(notes, coeff_UE3)]
    texte = "UE1 : " + str(moyennes[0]) + "\n"
    texte += "UE2 : " + str(moyennes[1]) + "\n"
    texte += "UE3 : " + str(moyennes[2]) + "\n\n"
    texte += decision(moyennes)
    resultat.config(text=texte)

fenetre = Tk()
fenetre.title("Validation du BUT 1")

frm = ttk.Frame(fenetre)
frm.grid()

matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]
saisies = []
for i, matiere in enumerate(matieres):
    ttk.Label(frm, text=matiere).grid(column=0, row=i, padx=10, pady=2)
    saisie = ttk.Entry(frm, width=10)
    saisie.grid(column=1, row=i, padx=10)
    saisies.append(saisie)

bouton = ttk.Button(frm, text="Valider", command=cal)
bouton.grid(column=2, row=0, padx=20)

resultat = ttk.Label(frm, text="")
resultat.grid(column=2, row=1, rowspan=6, padx=20)

fenetre.mainloop()
```
