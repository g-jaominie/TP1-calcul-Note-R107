```python
matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

liste_UE1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
liste_UE2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
liste_UE3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]
liste_notes = [13, 4, 11, 7, 5, 7, 2, 15, 16, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]

coeffs = {
    "UE1": dict(zip(matieres, liste_UE1)),
    "UE2": dict(zip(matieres, liste_UE2)),
    "UE3": dict(zip(matieres, liste_UE3)),
}
notes = dict(zip(matieres, liste_notes))

def calcul_moyenne(ue):
    somme_notes = 0
    somme_coeff = 0
    for matiere in notes:
        somme_notes += notes[matiere] * coeffs[ue][matiere]
        somme_coeff += coeffs[ue][matiere]
    return round(somme_notes / somme_coeff, 2)

def decision(moyennes):
    nb_valides = 0
    for ue in moyennes:
        if moyennes[ue] >= 10:
            nb_valides += 1
    petite = min(moyennes.values())
    if nb_valides == 3:
        return "Année validée, passage en BUT 2"
    elif nb_valides == 2 and petite >= 8:
        return "Année non validée, passage en BUT 2 autorisé (UE à valider en BUT 2)"
    elif nb_valides == 2:
        return "Année non validée, passage non autorisé (redoublement possible)"
    else:
        return "Année non validée, passage en BUT 2 exclu"

moyennes = {}
for ue in coeffs:
    moyennes[ue] = calcul_moyenne(ue)
    print("Moyenne", ue, ":", moyennes[ue])

print(decision(moyennes))
```
