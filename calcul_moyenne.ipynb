```python
matieres = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

def calcul_moy_ue(nom_ue):
    somme_notes = 0
    somme_coef = 0
    for matiere in matieres:
        note = float(input(nom_ue + " - note de " + matiere + " : "))
        coef = float(input(nom_ue + " - coefficient de " + matiere + " (0 si non concerné) : "))
        somme_notes += note * coef
        somme_coef += coef
    return somme_notes / somme_coef

moy1 = calcul_moy_ue("UE1")
moy2 = calcul_moy_ue("UE2")
moy3 = calcul_moy_ue("UE3")
print("Moyennes :", round(moy1, 2), round(moy2, 2), round(moy3, 2))

nb_valides = 0
for m in [moy1, moy2, moy3]:
    if m >= 10:
        nb_valides += 1
petite = min(moy1, moy2, moy3)

if nb_valides == 3:
    print("Année validée, passage en BUT 2")
elif nb_valides == 2 and petite >= 8:
    print("Année non validée, passage en BUT 2 autorisé (UE à valider en BUT 2)")
elif nb_valides == 2:
    print("Année non validée, passage non autorisé (redoublement possible)")
else:
    print("Année non validée, passage en BUT 2 exclu")
```
