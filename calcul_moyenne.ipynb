```python
coeff_UE1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
coeff_UE2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
coeff_UE3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

tb_notes = [13, 4, 11, 7, 5, 7, 2, 15, 16, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]

def calcul_moyenne(notes, coeff):
    somme_notes = 0
    somme_coeff = 0
    for i in range(len(notes)):
        somme_notes += notes[i] * coeff[i]
        somme_coeff += coeff[i]
    return round(somme_notes / somme_coeff, 2)

def decision(moyennes):
    nb_valides = 0
    for m in moyennes:
        if m >= 10:
            nb_valides += 1
    if nb_valides == 3:
        return "Année validée, passage en BUT 2"
    elif nb_valides == 2 and min(moyennes) >= 8:
        return "Année non validée, passage en BUT 2 autorisé (UE à valider en BUT 2)"
    elif nb_valides == 2:
        return "Année non validée, passage non autorisé (redoublement possible)"
    else:
        return "Année non validée, passage en BUT 2 exclu"

moyennes = [calcul_moyenne(tb_notes, coeff_UE1),
            calcul_moyenne(tb_notes, coeff_UE2),
            calcul_moyenne(tb_notes, coeff_UE3)]

print("Moyennes UE1, UE2, UE3 :", moyennes)
print(decision(moyennes))
```
