```python
import numpy as np
import matplotlib.pyplot as plt

coeff_UE1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
coeff_UE2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
coeff_UE3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

tb_notes = [13, 4, 11, 7, 5, 7, 2, 15, 16, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]      

def moy(coeff):
    somme_notes = 0
    somme_coefficients = 0
    for i in range(len(coeff)):
        somme_notes += tb_notes[i] * coeff[i]
        somme_coefficients += coeff[i]
    
    if somme_coefficients == 0:
        return 0
    return round(somme_notes / somme_coefficients, 3)

def couleur(m):
    if m >= 10:
        return "green"
    elif 8 <= m < 10:
        return "orange"
    else:
        return "red"

# Calcul des moyennes des trois UE
m1 = moy(coeff_UE1)
m2 = moy(coeff_UE2)
m3 = moy(coeff_UE3)

couleurs = [couleur(m1), couleur(m2), couleur(m3)]

print(f"la moyenne de l'UE1 est {m1}, de l'UE2 est {m2}, de l'UE3 est {m3}")

plt.bar(["UE1", "UE2", "UE3"], [m1, m2, m3], color=couleurs)
```
